```python
import tiktoken

# Văn bản cần token hóa
text = "Trí tuệ nhân tạo đang thay đổi cách chúng ta học tập"

# Sử dụng bộ mã hóa phổ biến của các mô hình OpenAI
encoding = tiktoken.get_encoding("o200k_base")

# Chuyển văn bản thành danh sách mã token
token_ids = encoding.encode(text)

print("Văn bản gốc:")
print(text)

print("\nDanh sách mã token:")
print(token_ids)

print("\nSố lượng token:")
print(len(token_ids))

print("\nChi tiết từng token:")
for index, token_id in enumerate(token_ids, start=1):
    token_text = encoding.decode([token_id])

    print(
        f"Token {index:2}: "
        f"ID = {token_id:6} | "
        f"Nội dung = {token_text!r}"
    )
```
